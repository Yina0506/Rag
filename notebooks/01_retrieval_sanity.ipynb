# 01 — Retrieval sanity check

Phase 1 exit criteria (`docs/03-phase-1-retrieval.md`): run the 10 seed claims
in `data/eval/test_claims.jsonl` through `pipeline.retrieve_candidates` and
eyeball that results look relevant. Not a formal eval — that's `09-evaluation.md`.

```python
import json
from pathlib import Path

from rag.pipeline import retrieve_candidates

claims_path = Path("../data/eval/test_claims.jsonl")
claims = [json.loads(line) for line in claims_path.read_text().splitlines()]
claims
```

```python
for row in claims:
    candidates = retrieve_candidates(row["claim"], k=5)
    print(row["claim"])
    for c in candidates:
        print(f"  {c.score:.3f}  {c.paper.title}")
```
